```python
import os
import re

def extract_imports(file_path):
    with open(file_path, 'r') as file:
        content = file.read()
    imports = re.findall(r'^\s*(?:import|from)\s+([a-zA-Z0-9_.]+)', content, re.MULTILINE)
    return set(imports)

def is_local_module(module_name, root_dir):
    module_path = module_name.replace('.', '/')
    return os.path.exists(os.path.join(root_dir, f'{module_path}.py')) or \
           os.path.exists(os.path.join(root_dir, module_path))

def generate_requirements(root_dir):
    requirements = set()
    for root, _, files in os.walk(root_dir):
        for file in files:
            if file.endswith('.py'):
                file_path = os.path.join(root, file)
                imports = extract_imports(file_path)
                requirements.update(imports)

    # Filter out local modules
    requirements = {req for req in requirements if not is_local_module(req, root_dir)}

    with open('requirements.txt', 'w') as req_file:
        for req in sorted(requirements):
            req_file.write(f'{req}\n')

    print('requirements.txt generated.')

if __name__ == '__main__':
    generate_requirements('app')  # Change 'app' to your specific folder
```

```python
import os
import re

def extract_imports(file_path):
    with open(file_path, 'r') as file:
        content = file.read()
    imports = re.findall(r'^\s*(?:import|from)\s+([a-zA-Z0-9_.]+)', content, re.MULTILINE)
    return set(imports)

def is_local_module(module_name, root_dirs):
    module_path = module_name.replace('.', '/')
    for root_dir in root_dirs:
        if os.path.exists(os.path.join(root_dir, f'{module_path}.py')) or \
           os.path.exists(os.path.join(root_dir, module_path)):
            return True
    return False

def gather_requirements(root_dirs):
    requirements = set()
    for root_dir in root_dirs:
        for root, _, files in os.walk(root_dir):
            for file in files:
                if file.endswith('.py'):
                    file_path = os.path.join(root, file)
                    imports = extract_imports(file_path)
                    requirements.update(imports)

    # Filter out local modules
    requirements = {req for req in requirements if not is_local_module(req, root_dirs)}

    with open('requirements.txt', 'w') as req_file:
        for req in sorted(requirements):
            req_file.write(f'{req}\n')

    print('requirements.txt generated.')

if __name__ == '__main__':
    # List all directories containing Python files
    root_dirs = ['bcbs', 'medicaid', 'medicare','optum','Vytalize','wellpoint']  # Add all relevant folders here
    gather_requirements(root_dirs)
```
